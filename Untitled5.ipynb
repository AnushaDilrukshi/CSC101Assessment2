```python
import random

def randomize_string(input_string):
    """Randomly arranges characters using a loop"""
    chars = list(input_string)
    result = ""
    # Use loop to randomly arrange
    while chars:
        rand_index = random.randint(0, len(chars) - 1)
        result += chars.pop(rand_index)
    return result

# Main program with exception handling
try:
    user_input = input("Enter a string: ")

    if not user_input:
        raise ValueError("String cannot be empty!")

    randomized = randomize_string(user_input)
    print(f"Randomized string: {randomized}")

except ValueError as ve:
    print(f"Error: {ve}")

except Exception as e:
    print(f"An unexpected error occurred: {e}")

finally:
    print("Program completed.")
```

```python
from decimal import Decimal, getcontext

# Set precision for currency
getcontext().prec = 10

class CurrencyCalculator:
    def __init__(self):
        print("Currency Calculator Initialized")

    def add(self, a, b):
        return Decimal(a) + Decimal(b)

    def subtract(self, a, b):
        return Decimal(a) - Decimal(b)

    def multiply(self, a, b):
        return Decimal(a) * Decimal(b)

    def divide(self, a, b):
        if Decimal(b) == 0:
            raise ZeroDivisionError("Cannot divide by zero!")
        return Decimal(a) / Decimal(b)

# Main program with exception handling
try:
    calc = CurrencyCalculator()

    num1 = input("Enter first amount: ")
    num2 = input("Enter second amount: ")

    if not num1 or not num2:
        raise ValueError("Amount cannot be empty!")

    print(f"\nAddition: {calc.add(num1, num2)}")
    print(f"Subtraction: {calc.subtract(num1, num2)}")
    print(f"Multiplication: {calc.multiply(num1, num2)}")
    print(f"Division: {calc.divide(num1, num2)}")

except ValueError as ve:
    print(f"Value Error: {ve}")

except ZeroDivisionError as ze:
    print(f"Math Error: {ze}")

except Exception as e:
    print(f"Unexpected error: {e}")

finally:
    print("\nProgram completed.")
```
