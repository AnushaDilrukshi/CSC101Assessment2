## Problem 1: Tip Calculator

### Step 1: Understand the Problem

This program helps restaurant staff calculate tips easily. The staff enter the bill amount and tip percentage. The program calculates the tip amount and the total amount the customer needs to pay.

### Step 2: Identify Inputs and Outputs

**INPUTS:**
- Bill amount: $45.80 (Number)
- Tip percentage: 15% (Number)

**OUTPUTS:**
- Tip amount in dollars ($)
- Total amount to pay in dollars ($)
- Both amounts displayed with two decimal places.

### Step 3: Work It By Hand

1. First, I multiply $45.80 by 0.15 to calculate the tip.

   Tip = $45.80 × 0.15 = $6.87

2. Next, I add the tip to the original bill.

   Total = $45.80 + $6.87 = $52.67

3. The final amount to charge is $52.67.

4. I would tell the staff member that the tip is $6.87 and the total bill is $52.67.

## Problem 2: Late Fee Calculator

### Step 1: Understand the Problem

This program helps library staff calculate fines for late books. The librarian enters the number of days the book is overdue. The program calculates the fine at 0.50 dollars per day, with a maximum fine of 10 dollars.

### Step 2: Identify Inputs and Outputs

**INPUTS:**
- Number of days the book is overdue (Whole number).

**OUTPUTS:**
- The late fee amount in dollars.
- If the calculated fine is more than 10 dollars, display "Maximum fine reached".
- If the book is not late, display a fine of 0.00 dollars.

### Step 3: Work It By Hand

**Case A: Book is 5 days late**

- My calculation: 5 × $0.50 = $2.50
- Fine: $2.50

**Case B: Book is 25 days late**

- My calculation: 25 × $0.50 = $12.50
- Fine: $10.00
- Special note: The calculated fine is 12.50 dollars, but the maximum allowed fine is 10 dollars. Therefore, the librarian charges only 10 dollars.




**Case C: Book is 0 days late**

- My calculation: 0 × $0.50 = $0.00
- Fine: $0.00
