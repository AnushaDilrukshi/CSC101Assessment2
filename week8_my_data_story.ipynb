```python
# Cell 1: Upload and Read my_data.txt
from google.colab import files
import os

# Check if file already exists, if not ask to upload
if not os.path.exists("my_data.txt"):
    print("Please upload your my_data.txt file")
    uploaded = files.upload()
else:
    print("my_data.txt found!")

# Read and display the file
with open("my_data.txt", "r", encoding="utf-8") as f:
    content = f.read()

print("--- Your Goals ---")
print(content)
print("\n-------------------")
print(f"Total characters: {len(content)}")
print(f"Total lines: {len(content.splitlines())}")
print(f"Total words: {len(content.split())}")
```
